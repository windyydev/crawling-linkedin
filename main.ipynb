```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from time import sleep
import random
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
#Step 1: Login Linkin
driver = webdriver.Chrome()
url = 'https://www.linkedin.com/login'
#url = 'https://cellphones.com.vn/'
driver.get(url)
sleep(3)

#read file txt
credential = open('login_credential.txt')
line = credential.readlines()
username = line[0]
password = line[1]

# import email
email_field = driver.find_element(By.ID, 'username')
email_field.send_keys(username)
sleep(2)

#import password
password_field = driver.find_element(By.NAME, 'session_password')
password_field.send_keys(password)
sleep(2)

# click login button
login_button = driver.find_element(By.CSS_SELECTOR, '#organic-div > form > div.login__form_action_container > button')
login_button.click()
sleep(random.uniform(3,6))
```

```python
#Step 2
#Search cái cần tìm
search_field = driver.find_element(By.CSS_SELECTOR, 'input[data-testid="typeahead-input"]')
search_field.send_keys('Software Engineer')
search_field.send_keys(Keys.RETURN)

wait = WebDriverWait(driver, 10)
people_tab = wait.until(EC.element_to_be_clickable((By.XPATH,'//label[contains(text(),"Jobs")]')))
people_tab.click()

# wait = WebDriverWait(driver, 10)
# people_tab2 = wait.until(EC.element_to_be_clickable((By.XPATH, '//label[contains(text(),"Backend")]')))
# people_tab2.click()
```
